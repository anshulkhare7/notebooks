```python
DRIVE_DATA_PATH = '/content/drive/MyDrive/Predicto_GPT_Taining_images'
```

```python
import os

def rename_jpg_to_jpeg(directory, subfolders):
    for subfolder in subfolders:
        subfolder_path = os.path.join(directory, subfolder)
        if os.path.isdir(subfolder_path):
            for root, _, files in os.walk(subfolder_path):
                for file in files:
                    if file.lower().endswith(".jpg"):
                        old_path = os.path.join(root, file)
                        new_path = os.path.join(root, os.path.splitext(file)[0] + ".jpeg")
                        try:
                            os.rename(old_path, new_path)
                            print(f"Renamed: {old_path} -> {new_path}")
                        except Exception as e:
                            print(f"Error renaming {old_path}: {e}")
        else:
            print(f"Warning: Subfolder '{subfolder}' not found in '{directory}'")

subfolders_to_process = ['50', '25', '200', '175', '350', '400', '300', '250', '450', '75', '0', '130', '100', 'Invalid']
rename_jpg_to_jpeg(DRIVE_DATA_PATH, subfolders_to_process)
print("✅ Renaming complete.")
```
